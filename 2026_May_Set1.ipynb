```python
"""
Q1 - Write ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠a function traffic_signal(vehicle_type, signal, emergency) that tells a road user what to do.

If emergency is True and vehicle_type is "ambulance", return "Proceed Carefully".
Otherwise it depends on signal:
"red": return "Stop".
"yellow": return "Slow Down".
"green": return "Cross" if vehicle_type is "pedestrian", and "Go" for any other vehicle type.
For any other signal, return "Invalid Signal".
NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
traffic_signal("car", "red", False) -> 'Stop'
traffic_signal("pedestrian", "green", False) -> 'Cross'
Your task
Implement this function:

def traffic_signal(vehicle_type: str, signal: str, emergency: bool) -> str:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: vehicle_type, signal, emergency.

Example

Input
"car"
"red"
False

Output
'Stop'

"""


def traffic_signal(vehicle_type: str, signal: str, emergency: bool) -> str:
    if signal not in ("green", "red", "yellow"):
        return "Invalid Signal"
    elif emergency and vehicle_type == "ambulance":
        return "Proceed Carefully"
    elif signal == "red":
        return "Stop"
    elif signal == "yellow":
        return "Slow Down"
    elif signal == "green":
        if vehicle_type == "pedestrian":
            return "Cross"
        else:
            return "Go"
```

```python
"""
Q2 - The ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠list words has exactly four strings. Write a function join_even_index_strings(words) that returns one string made by joining the strings at the even indices, which are index 0 and index 2, in this order, with nothing in between.

NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
join_even_index_strings(["apple", "banana", "cherry", "mango"]) -> 'applecherry'
join_even_index_strings(["red", "green", "blue", "yellow"]) -> 'redblue'
Your task
Implement this function:

def join_even_index_strings(words: list) -> str:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: words.

Example
Input

["apple", "banana", "cherry", "mango"]

Output
'applecherry'

"""

def join_even_index_strings(words:list) -> str:
    word = words[0]
    for i in range(1, len(words)):
        if i % 2 == 0:
            word += words[i]

    return word


```

```python
"""
Q3 - Write ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠a function make_dict_from_elems_in_index(keys, values, index) that returns a dictionary with just one entry: the key is the element of keys at position index and the value is the element of values at position index. The index is valid for both lists, and it can be negative. The two lists can have different lengths.

NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
make_dict_from_elems_in_index(['apple', 'banana', 'cherry'], [10, 20, 30, 40], 1) -> {'banana': 20}
make_dict_from_elems_in_index(['apple', 'banana', 'cherry'], [10, 20, 30, 40], -1) -> {'cherry': 40}
Your task
Implement this function:

def make_dict_from_elems_in_index(keys: list, values: list, index: int) -> dict:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: keys, values, index.

Example
Input

['apple', 'banana', 'cherry']
[10, 20, 30, 40]
1
Output

{'banana': 20}


"""

def make_dict_from_elems_in_index(keys:list, values: list, index:int) -> dict:
    return {keys[index]: values[index]}
```

```python
"""
Q1 - Write ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠a function find_zero_pairs(nums) that takes a list of integers and returns a set of the pairs of numbers of the list whose sum is 0. Each pair is a tuple with its smaller number first, and a pair appears only once in the set, however many times it can be found in the list. A number can be used twice in a pair only if it occurs at least twice in the list, so 0 pairs with itself only if the list has at least two zeros.

NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
find_zero_pairs([1, -1, 2, -2, 3]) -> {(-2, 2), (-1, 1)}
find_zero_pairs([0, 0, 0]) -> {(0, 0)}
Your task
Implement this function:

def find_zero_pairs(nums: list) -> set:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: nums.

Example
Input

[1, -1, 2, -2, 3]
Output

{(-2, 2), (-1, 1)}



"""


def find_zero_pairs(lst:list) -> set:
    seen = set()
    for i in range(len(lst)):
        for j in range(i+1, len(lst)):
            if lst[i] + lst[j] == 0:
                pair = tuple(sorted((lst[i], lst[j])))
                seen.add(pair)

    return seen

```

```python
"""
Read n lines ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠of text and print how many of all their characters are digits (0 to 9).

Input Format
The first line contains the integer n.
Each of the next n lines is a line of text.

Output Format
Print one integer, the number of digit characters.

NOTE: This is an I/O type question, you need to write the whole code for taking input and printing the output.

Your task
Write a complete program: read the input with input() and print the result with print().

Example

Input

3
hello 123
world 45
no digits here

Output

5
"""

n = int(input())
count = 0
for _ in range(n):
    for ch in input():
        if ch.isdigit():
            count += 1

print(count)
```

```python
"""
Q1 - data is ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠a dictionary from the names of users to the lists of their workout durations in minutes. Write a function filter_workouts(data, category) that returns the set of the names of the users who belong to the category:

"fitness_enthusiast": the average duration is at least 90.
"regular_user": the average duration is at least 45 and less than 90.
"consistent_workout": every duration is at least 30.
"balanced_routine": the longest duration minus the shortest duration is at most 20.
For any other category, return an empty set.

NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
filter_workouts({"Aditi": [100, 95, 90], "Bharat": [50, 60, 55], "Chetan": [120, 130, 125], "Divya": [35, 40, 45], "Esha": [20, 55, 25]}, "fitness_enthusiast") -> {'Aditi', 'Chetan'}
filter_workouts({"Aditi": [100, 95, 90], "Bharat": [50, 60, 55], "Chetan": [120, 130, 125], "Divya": [35, 40, 45], "Esha": [20, 55, 25]}, "consistent_workout") -> {'Aditi', 'Bharat', 'Chetan', 'Divya'}
Your task
Implement this function:

def filter_workouts(data: dict, category: str) -> set:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: data, category.

Example->

Input
{"Aditi": [100, 95, 90], "Bharat": [50, 60, 55], "Chetan": [120, 130, 125], "Divya": [35, 40, 45], "Esha": [20, 55, 25]}
"fitness_enthusiast"

Output
{'Aditi', 'Chetan'}
"""
def filter_workouts(data: dict, category: str) -> set:
    if category == "fitness_enthusiast":
        return {k for k, v in data.items() if sum(v) / len(v) >= 90}
    elif category == "regular_user":
        return {k for k, v in data.items() if 45 <= sum(v) / len(v) < 90}
    elif category == "balanced_routine":
       
        return {k for k, v in data.items() if all(x > 30 for x in v)}
    elif category == "consistent_workout":
        return {k for k, v in data.items() if all(x >= 30 for x in v)}
    return set()
```

```python
"""
Q2 - Read ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠an integer n (at least 2) and print a hollow square of n rows, each with n columns, using the character #. Only the border of the square is filled with #; the inside is made of spaces. A row does not end with spaces other than the ones before its last #.

Input Format
A single integer n.

Output Format
The hollow square. For n = 4 it is

####
#  #
#  #
####
NOTE: This is an I/O type question, you need to write the whole code for taking input and printing the output.

Your task
Write a complete program: read the input with input() and print the result with print().

Example
Input

4
Output

####
#  #
#  #
####


"""

n = int(input())

for i in range(n):
    if i == 0 or i == n - 1:
        print("#" * n)
    else:
        print("#" + " " * (n - 2) + "#")
```
