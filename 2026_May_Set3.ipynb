```python
"""
Write ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠a function is_tens_digit_greater(num) that takes a positive two-digit integer and returns True if its tens digit is strictly greater than its ones digit, and False otherwise.

NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
is_tens_digit_greater(54) -> True
is_tens_digit_greater(33) -> False
Your task
Implement this function:

def is_tens_digit_greater(num: int) -> bool:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: num.

Example
Input

54
Output

True
"""


def is_tens_digit_greater(num: int) -> bool:
    n = str(num)[0]
    n1 = str(num)[-1]
    return int(n) > int(n1)
```

```python
"""
Write ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠a function reverse_each_word(words) that takes a list of strings and returns a new list in which every string is written backwards, with the strings in the same order.

NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
reverse_each_word(["apple", "cat", "python"]) -> ['elppa', 'tac', 'nohtyp']
reverse_each_word(["hi", "go"]) -> ['ih', 'og']
Your task
Implement this function:

def reverse_each_word(words: list) -> list:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: words.

Example
Input

["apple", "cat", "python"]
Output

['elppa', 'tac', 'nohtyp']
"""
def reverse_words(words:list) -> list:
    return [word[::-1] for word in words]
```

```python
"""
Read n entries. Each ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠entry is a line with an integer and a string (with no spaces), separated by a space. Find the entry whose string is the longest and print its integer on one line and its string on the next. If several entries tie for the longest string, use the one which appears first in the input.

Input Format
The first line contains the integer n.
Each of the next n lines contains an integer and a string, separated by a space.

Output Format
Two lines: the integer and the string of the chosen entry.

NOTE: This is an I/O type question, you need to write the whole code for taking input and printing the output.

Your task
Write a complete program: read the input with input() and print the result with print().

Example
Input

4
1 apple
2 banana
3 kiwi
4 mango
Output

2
banana
"""
n = int(input())
entries = []
for _ in range(n):
    num_str, word = input().split()
    entries.append((int(num_str), word))
best_num, best_word = entries[0]
for num, word in entries[1:]:
    if len(word) > best_word:
        best_word = word
        best_num = num

print(best_num)
print(best_word)
```

```python
"""
Q1 - data is ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠a dictionary from the names of customers to the lists of their monthly spending amounts. Write a function filter_customers(data, command) that returns the set of the names of the customers who match the command:

"high_spender": the average spending is at least 50000.
"moderate_spender": the average spending is at least 20000 and less than 50000.
"consistent_spender": the spending of every month is at least 10000.
"stable_spending": the largest monthly amount minus the smallest one is at most 5000.
For any other command, return an empty set.

NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
filter_customers({"Akhil": [45000, 55000, 50000], "Bina": [12000, 18000, 15000], "Charan": [70000, 65000, 72000], "Divya": [25000, 30000, 28000], "Esha": [15000, 16000, 15500]}, "high_spender") -> {'Akhil', 'Charan'}
filter_customers({"Akhil": [45000, 55000, 50000], "Bina": [12000, 18000, 15000], "Charan": [70000, 65000, 72000], "Divya": [25000, 30000, 28000], "Esha": [15000, 16000, 15500]}, "stable_spending") -> {'Divya', 'Esha'}
Your task
Implement this function:

def filter_customers(data: dict, command: str) -> set:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: data, command.

Example
Input

{"Akhil": [45000, 55000, 50000], "Bina": [12000, 18000, 15000], "Charan": [70000, 65000, 72000], "Divya": [25000, 30000, 28000], "Esha": [15000, 16000, 15500]}
"high_spender"
Output

{'Akhil', 'Charan'}
"""
def filter_customers(data: dict, command: str) -> set:
    res = set()
    for name, values in data.items():
        if not values:
            continue

        avg = sum(values) / len(values)

        if command == "high_spender":
            if avg >= 50000:
                res.add(name)
        elif command == "moderate_spender":
            if 20000 <= avg < 50000:
                res.add(name)
        elif command == "consistent_spender":
            if all(vals >= 10000 for vals in values):
                    res.add(name)
        elif command == "stable_spending":
            if max(values) - min(values) <= 5000:
                res.add(name)
    return res
```

```python
"""
Read ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠an integer n (at least 3) and print an X made of * inside a grid of n rows and n columns. A * is printed on both diagonals of the grid, and spaces everywhere else. In the middle row of a grid with an odd n the two diagonals meet, and only one * is printed there. Rows do not end with spaces.

Input Format
A single integer n.

Output Format
The X pattern. For n = 5 it is

*   *
 * *
  *
 * *
*   *
NOTE: This is an I/O type question, you need to write the whole code for taking input and printing the output.

Your task
Write a complete program: read the input with input() and print the result with print().

Example
Input

5
Output

*   *
 * *
  *
 * *
*   *

Code Editor

"""

n = int(input())

for r in range(n):
    row = ""
    for c in range(n):
        if r == c or r + c == n - 1:
            row += "*"
        else:
            row += " "
    print(row.rstrip())
```
