```python
"""
Q2 - Write ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠a function signed_abs_diff_of_middle_digits(n) that looks at the digits of n (without its sign). If there is an odd number of digits, take the single middle digit. If there is an even number of digits, take the absolute difference of the two middle digits. The function returns that number, with a minus sign if n is negative.

NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
signed_abs_diff_of_middle_digits(12345) -> 3
signed_abs_diff_of_middle_digits(-987) -> -8
Your task
Implement this function:

def signed_abs_diff_of_middle_digits(n: int) -> int:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: n.

Example
Input

12345
Output

3


"""

def signed_abs_diff_of_middle_digits(n: int) -> int:
    s = str(abs(n))
    mid = len(s) // 2

    if len(s) % 2 == 1:
        val = int(s[mid])
    else:
        val = abs(int(s[mid - 1]) - int(s[mid]))

    return -val if n < 0 else val
```

```python
"""
Q3 - Write ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠a function is_decreasing_number(n: int) -> bool that checks if a given 4-digit number is decreasing, meaning each digit is strictly less than the digit before it (from left to right).

The function should return True if the number meets the criteria and False otherwise.

Constraints
The input number will be a 4-digit number, which means it will be in the range of 1000 to 9999.

Example
>>> is_decreasing_number(4321)
True
>>> is_decreasing_number(4312)
False
>>> is_decreasing_number(9876)
True
>>> is_decreasing_number(1111)
False
>>> is_decreasing_number(3210)
True
Your task
Implement this function:

def is_decreasing_number(n: int) -> bool:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: n.

Example
Input

4321
Output

True
"""
def is_decreasing_number(n: int) -> bool:
    digits = str(n)
    return digits[0] > digits[1] > digits[2] > digits[3]
```

```python
"""
Q1 - A ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠run is a stretch of equal elements that sit next to each other in the list. Write a function remove_k_consecutive_duplicates(nums, k) that returns a new list in which every run of k or more equal elements has been removed completely. The elements of shorter runs stay, in their original order.

NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
remove_k_consecutive_duplicates([1, 1, 1, 2, 2, 3, 3, 3, 3], 3) -> [2, 2]
remove_k_consecutive_duplicates([1, 1, 2, 2, 2, 1, 1], 2) -> []
Your task
Implement this function:

def remove_k_consecutive_duplicates(nums: list, k: int) -> list:
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: nums, k.

Example
Input

[1, 1, 1, 2, 2, 3, 3, 3, 3]
3
Output

[2, 2]
"""


def remove_k_consecutive_duplicates(nums: list, k: int) -> list:
    res = []
    i = 0
    n = len(nums)

    while i < n:
        j = i
        # Move j forward while elements match nums[i]
        while j < n and nums[j] == nums[i]:
            j += 1

        run_length = j - i
        if run_length < k:
            res.extend(nums[i:j])   # Keep the whole run if shorter than k

        i = j   # Jump straight to the start of the next run

    return res
```

```python
"""
Q2 - Read n movie ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠lists. Each list is given by two lines: the first line has the name of the list, a semicolon, and the names of the movies separated by commas; the second line has the watch time of every movie of the list, in minutes, separated by commas.

Print one line for each list, in the input order, with the name of the list, a colon, a space and the total watch time of the list. Then print one more line: Unique movies across all lists:, a space, and the names of all the different movies of all the lists in alphabetical order, separated by a comma and a space.

Input Format
The first line contains the integer n.
Then come n pairs of lines, as described above.

Output Format
The total watch time of each list, followed by the line of the unique movies.

NOTE: This is an I/O type question, you need to write the whole code for taking input and printing the output.

Your task
Write a complete program: read the input with input() and print the result with print().

Example
Input

3
Watchlist1;Inception,Dunkirk,Tenet
120,106,150
Watchlist2;Interstellar,Tenet
169,150
Watchlist3;Inception,Joker
120,122
Output

Watchlist1: 376
Watchlist2: 319
Watchlist3: 242
Unique movies across all lists: Dunkirk, Inception, Interstellar, Joker, Tenet
"""

n = int(input())
unique_movies = set()

for _ in range(n):
    name, movies_str = input().split(";")
    movies = movies_str.split(",")
    times = list(map(int, input().split(",")))

    print(f"{name}: {sum(times)}")
    unique_movies.update(movies)

print("Unique movies across all lists:", ", ".join(sorted(unique_movies)))
```

```python
"""
Q1 - records is ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠a list of tuples (student_id, score, subject). Write a function process_records(records, operation, subject=None) that does what operation asks for:

"count_records": return the number of records.
"average_by_subject": return the average score of the records of the given subject.
"top_student_by_subject": return the student_id of the record with the highest score among the records of the given subject. If several records tie, the first one in the list wins.
"pass_count": return how many records of the given subject have a score of at least 60.
For one of the last three operations, if subject is None or no record has that subject, return an empty dictionary {}. For any other operation, return {} as well. A number you return is compared after rounding it to 4 decimal places.

NOTE: This is a function type question, you don't have to take input or print the output, just have to complete the required function definition.

Example
process_records([("S1", 85, "Math"), ("S2", 72, "Math"), ("S3", 91, "Science"), ("S4", 55, "Math"), ("S5", 65, "Science")], "count_records") -> 5
process_records([("S1", 85, "Math"), ("S2", 72, "Math"), ("S3", 91, "Science"), ("S4", 55, "Math"), ("S5", 65, "Science")], "average_by_subject", "Math") -> 70.6667
Your task
Implement this function:

def process_records(records: list, operation: str, subject=None):
Write the function only — the arguments are read for you and the return value is printed automatically.

Arguments arrive as one Python literal per line, in this order: records, operation, subject.

Example
Input

[("S1", 85, "Math"), ("S2", 72, "Math"), ("S3", 91, "Science"), ("S4", 55, "Math"), ("S5", 65, "Science")]
"count_records"
Output - 5
"""

def process_records(records: list, operation: str, subject=None):
    if operation == "count_records":
        return len(records)

    if operation not in ("average_by_subject", "top_student_by_subject", "pass_count"):
        return {}

    if subject is None:
        return {}

    sub_records = [(sid, score) for sid, score, subj in records if subj == subject]
    if not sub_records:
        return {}

    if operation == "average_by_subject":
        return sum(score for _, score in sub_records) / len(sub_records)

    elif operation == "top_student_by_subject":
        best_sid, best_score = sub_records[0]
        for sid, score in sub_records[1:]:
            if score > best_score:
                best_sid, best_score = sid, score
        return best_sid

    elif operation == "pass_count":
        return sum(1 for _, score in sub_records if score >= 60)
```

```python
"""
Q2 - Read ⁠​‌‌​‌​​‌​‌​‌‌​​‌‌‌​‌‌‌​​​‌‌​​‌‌​⁠an integer n (from 1 to 9) and print an hourglass of 2n - 1 rows. Each row is a palindrome of numbers separated by single spaces: the row for k is 1 2 ... k ... 2 1, counting up to k and back down to 1. The rows go from k = n down to k = 1 for the upper half, and then from k = 2 up to k = n for the lower half. Each row is shifted to the right by two spaces for every step that its k is below n, so that the hourglass is centred.

Input Format
A single integer n.

Output Format
The hourglass, as described above. For n = 3 it is

1 2 3 2 1
  1 2 1
    1
  1 2 1
1 2 3 2 1
NOTE: This is an I/O type question, you need to write the whole code for taking input and printing the output.

Your task
Write a complete program: read the input with input() and print the result with print().

Example
Input

3
Output

1 2 3 2 1
  1 2 1
    1
  1 2 1
1 2 3 2 1

"""

n = int(input())


ks = list(range(n, 0, -1)) + list(range(2, n + 1))

for k in ks:
    spaces = " " * (2 * (n - k))
    nums = list(range(1, k + 1)) + list(range(k - 1, 0, -1))
    print(spaces + " ".join(map(str, nums)))
```
